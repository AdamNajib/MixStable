```python
from Mixstable.simulations import evaluate_fit 
from Mixstable.visualization import plot_method_comparison
from Mixstable.em import em_fit_alpha_stable_mixture
from Mixstable.em_methode import (
    em_estimate_stable_recursive_ecf,
    em_estimate_stable_kernel_ecf,
    em_estimate_stable_weighted_ols,
    em_estimate_stable_from_cdf
)
from Mixstable.generate_sample import generate_synthetic_data
from Mixstable.r_interface import stats
import numpy as np

# Example data and frequency values
data = generate_synthetic_data()
h_n = stats.bw_SJ(data)[0]
print(f"Bande passante R (bw.SJ) : {h_n:.5f}")

# List of methods
methods = {
    "EM Estimated": em_fit_alpha_stable_mixture(data),
    "EM ECF recursive_ecf": em_estimate_stable_recursive_ecf(data),
    "EM ECF Kernel": em_estimate_stable_kernel_ecf(data),
    "EM ECF weighted_ols": em_estimate_stable_weighted_ols(data),
    "EM ECF from_cdf": em_estimate_stable_from_cdf(data)
}


# Example data
S = np.random.standard_t(df=1.5, size=500)
S_clean = S[~np.isnan(S)]
# Ensure methods dictionary does not contain invalid parameters
# Ensure methods dictionary contains valid parameters (tuples of floats)
methods_clean = {
    name: params for name, params in methods.items()
    if isinstance(params, tuple) and len(params) == 3 and all(isinstance(p, tuple) and len(p) == 4 and all(isinstance(x, (int, float)) for x in p) for p in params[:2]) and isinstance(params[2], (int, float))
}

# Evaluate fit and plot comparison
scores = evaluate_fit(S_clean, methods_clean, np.linspace(min(S_clean), max(S_clean), 1000))
plot_method_comparison(scores)
```

```python
from Mixstable.mcculloch import build_mcculloch_interpolators, generate_mcculloch_table
from Mixstable.simulations import compare_estimators_on_simulations  
import numpy as np

alpha_grid = np.linspace(0.1, 2.0, 100)
beta_grid = np.linspace(-1.0, 1.0, 100)
# Construire les interpolateurs
interp_alpha, interp_beta = build_mcculloch_interpolators(generate_mcculloch_table(alpha_grid, beta_grid, size=100_000, seed=42))

# Appeler la fonction principale avec les interpolateurs
df_results = compare_estimators_on_simulations(n_samples=1000,n_runs=30,interp_alpha=interp_alpha,interp_beta=interp_beta)

# Résumé des performances
summary = df_results.groupby("method").agg(["mean", "std"])
summary.to_latex("table_performance_summary.tex")
```
